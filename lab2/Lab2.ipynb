```python
import pandas as pd

# Load the original CSV file into a DataFrame
df_original = pd.read_csv("outputData/all_countries.csv")

# Display the first 5 rows of the DataFrame to confirm
print(df_original.head())
```

```python
# Create a pivot table with countries as rows, activities as columns,
# and hours per day as the values.
df_pivoted = df_original.pivot_table(
    index='countryISO3',
    columns='Subcategory',
    values='hoursPerDayCombined',
    aggfunc='sum'
).reset_index()

# Replace any missing values (NaN) with 0.
df_pivoted_filled = df_pivoted.fillna(0)

# Create a new column 'total_hours' that sums all the hours for each country.
# We slice the DataFrame to exclude the 'countryISO3' column before summing.
df_pivoted_filled['total_hours'] = df_pivoted_filled.iloc[:, 1:].sum(axis=1)

# Display the first few rows of the new DataFrame to show the result.
print("DataFrame with 'total_hours' column:")
print(df_pivoted_filled)
```

```python
num_unique_countries = df_original['countryISO3'].nunique()

print(f"The number of unique countries in the dataset is: {num_unique_countries}")
```
